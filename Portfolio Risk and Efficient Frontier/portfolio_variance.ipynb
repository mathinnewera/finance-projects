**Q1. Portfolio Risks**

(a) Matrix formula for portfolio variance


$$
\sigma_p^2 = \mathbf{w}^T \boldsymbol{\Sigma} \mathbf{w}
$$ or

$$
\sigma_p^2 = \sum_{i=1}^{6} \sum_{j=1}^{6} w_i w_j \sigma_{ij}
$$

where

$$
\mathbf{w} =
\begin{pmatrix}
w_1 \\
w_2 \\
\vdots \\
w_6
\end{pmatrix},
\quad
\boldsymbol{\Sigma} =
\begin{pmatrix}
\sigma_{11} & \sigma_{12} & \cdots & \sigma_{16} \\
\sigma_{21} & \sigma_{22} & \cdots & \sigma_{26} \\
\vdots & \vdots & \ddots & \vdots \\
\sigma_{61} & \sigma_{62} & \cdots & \sigma_{66}
\end{pmatrix}
$$

(b) Function to Compute Variance for Multiple Portfolios

```python
import numpy as np

def portfolio_variance(weights, cov_matrix):
    weights = np.array(weights)
    return weights.T @ cov_matrix @ weights
```
