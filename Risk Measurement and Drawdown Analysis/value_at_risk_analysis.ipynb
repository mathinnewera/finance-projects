#**Q1. VaR Analysis**

**(a) Distribution of Portfolio Return $r_p$**

Gaussian VaR at $\alpha$ quantile of an investment $I$ is $\text{VaR}_{\alpha} = -(\sigma Z_{\alpha}+\mu)\times I$

Suppose two investments $A$ and $B$ have returns that follow normal distributions:  $r_A \sim N(\mu_A, \sigma_A^2)$ and $r_B \sim N(\mu_B, \sigma_B^2)$ where $\rho_{AB}$ is the correlation between $r_A$ and $r_B$.

The return of the portfolio is $r_p = 0.5 r_A + 0.5 r_B$

Since $r_A$ and $r_B$ are jointly normal, $r_p$ is normal with mean: $\mu_p = \mathbb{E}[r_p] = 0.5\mu_A + 0.5\mu_B = \frac{\mu_A + \mu_B}{2}$

and variance:
$$
\begin{aligned}
\sigma_p^2 &= \text{Var}(r_p) = (0.5)^2 \sigma_A^2 + (0.5)^2 \sigma_B^2 + 2(0.5)(0.5) \rho_{AB} \sigma_A \sigma_B \\
&= 0.25 \sigma_A^2 + 0.25 \sigma_B^2 + 0.5 \rho_{AB} \sigma_A \sigma_B \\
&= 0.25 \left( \sigma_A^2 + \sigma_B^2 + 2 \rho_{AB} \sigma_A \sigma_B \right)
\end{aligned}
$$

<br>

$ ⟹ \sigma_p = 0.5 \sqrt{\sigma_A^2 + \sigma_B^2 + 2 \rho_{AB} \sigma_A \sigma_B}$

<br>

Therefore, $r_p \sim N\left( \mu_p,\; \sigma_p^2 \right)$

**(b) $\text{VaR}_\alpha$ of the Portfolio**

For $I=1$ (per unit invested), $\text{VaR}_\alpha(r_p) = -\left( \sigma_p Z_{\alpha} + \mu_p \right)$

Substitute $\mu_p = 0.5\mu_A + 0.5\mu_B = \frac{\mu_A + \mu_B}{2}$ and $ \sigma_p = 0.5 \sqrt{\sigma_A^2 + \sigma_B^2 + 2 \rho_{AB} \sigma_A \sigma_B}$ :

${\text{VaR}_\alpha(r_p) = -0.5\left[ \sqrt{\sigma_A^2 + \sigma_B^2 + 2 \rho_{AB} \sigma_A \sigma_B} \; Z_{\alpha} \;+\; \mu_A + \mu_B \right]}$

**(c) Proving the Subadditivity Property**

Objective (show that):
$\text{VaR}_\alpha(r_p) \le \text{VaR}_\alpha(r_A) + \text{VaR}_\alpha(r_B)$

By definition: $\text{VaR}_\alpha(r_A) = -(\sigma_A Z_{\alpha} + \mu_A)$ and $\text{VaR}_\alpha(r_B) = -(\sigma_B Z_{\alpha} + \mu_B)$

$ ⇒ -(\sigma_p Z_{\alpha} + \mu_p) \le -(\sigma_A Z_{\alpha} + \mu_A) - (\sigma_B Z_{\alpha} + \mu_B)$

$
⇒\sigma_p Z_{\alpha} + \mu_p \ge (\sigma_A + \sigma_B) Z_{\alpha} + \mu_A + \mu_B $

<br>

Since $\mu_p = \frac{\mu_A + \mu_B}{2}$ and $Z_{\alpha} < 0$ for $\alpha < 0.5$, from the triangle inequality for standard deviations:  $\sigma_p = 0.5\sqrt{\sigma_A^2 + \sigma_B^2 + 2\rho_{AB}\sigma_A\sigma_B} \le 0.5(\sigma_A + \sigma_B)$

Since $Z_{\alpha} < 0$ $⇒\sigma_p Z_{\alpha} \ge 0.5(\sigma_A + \sigma_B) Z_{\alpha}$

Adding $\mu_p = 0.5(\mu_A + \mu_B) ⇒$ $
\sigma_p Z_{\alpha} + \mu_p \ge 0.5(\sigma_A + \sigma_B) Z_{\alpha} + 0.5(\mu_A + \mu_B)$

The RHS is exactly $\frac{1}{2}\left[ (\sigma_A + \sigma_B) Z_{\alpha} + \mu_A + \mu_B \right]$, therefore $\text{VaR}_\alpha(r_p) \le \frac{\text{VaR}_\alpha(r_A) + \text{VaR}_\alpha(r_B)}{2}$

For $\alpha < 0.5$ and typical positive means, $\text{VaR}_\alpha(r_A) + \text{VaR}_\alpha(r_B) \ge 0$, so $\frac{\text{VaR}_A + \text{VaR}_B}{2} \le \text{VaR}_A + \text{VaR}_B$

Hence  $\text{VaR}_\alpha(r_p) \le \text{VaR}_\alpha(r_A) + \text{VaR}_\alpha(r_B)$

(subadditivity holds for Gaussian VaR with $\alpha < 0.5$).
